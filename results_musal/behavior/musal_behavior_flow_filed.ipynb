```python
from matplotlib.colors import Normalize
import matplotlib.cm as cm
```

```python
def plot_flow_field_3d(traj, skip=1, normalize=False, length=1.0,
                       elev_azim=None, fig_w=6, fig_h=4, alpha=0.9, alpha_path = 0.2,
                       show_path=False, linewidth=0.6, cmap_name='viridis',
                       add_colorbar=True, arrow_length_ratio = 0.7):
    traj = np.asarray(traj)
    if traj.ndim != 2 or traj.shape[1] != 3 or traj.shape[0] < 2:
        raise ValueError("traj must be (T, 3) with T >= 2")

    # anchors and step vectors
    P = traj[:-1:skip]
    V = np.diff(traj, axis=0)[::skip]

    # color by step speed
    speed = np.linalg.norm(V, axis=1)
    p5, p95 = np.percentile(speed, [5, 95])
    norm = Normalize(vmin=p5, vmax=p95)
    cmap = cm.get_cmap(cmap_name)
    colors = cmap(norm(np.clip(speed, p5, p95)))

    # keep lengths unless normalize=True; otherwise cap long arrows at p95
    if normalize:
        V_plot = V
    else:
        scale = np.minimum(speed, p95) / (speed + 1e-12)
        V_plot = V * scale[:, None]

    # figure / view
    fig = plt.figure(figsize=(fig_w, fig_h))
    ax = fig.add_subplot(111, projection='3d')
    if elev_azim is not None:
        ax.view_init(elev=elev_azim[0], azim=elev_azim[1])

    # ONLY arrows (no scatter); arrows are colored by speed
    ax.quiver(P[:,0], P[:,1], P[:,2],
              V_plot[:,0], V_plot[:,1], V_plot[:,2],
              length=length, normalize=normalize,
              color=colors, linewidths=linewidth, alpha=alpha, arrow_length_ratio = arrow_length_ratio)

    if show_path:
        ax.plot(traj[:,0], traj[:,1], traj[:,2], lw=0.8, alpha=alpha_path, c = 'k')

    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")

    if add_colorbar:
        sm = cm.ScalarMappable(norm=norm, cmap=cmap); sm.set_array([])
        fig.colorbar(sm, ax=ax, shrink=0.75, pad=0.08, label='step speed')

    plt.tight_layout()
```
